```python
logging.basicConfig(filename='app.log', filemode='w',level=logging.DEBUG, format='%(asctime)s - %(levelname)s - %(message)s', datefmt='%Y-%m-%d %H:%M:%S')
def add(x, y):
    logging.debug(f'Adding {x} and {y}')
    return x + y
add(5, 3)
```

```python
import logging
logger1 = logging.getLogger("module1")
logger1.setLevel(logging.DEBUG)
logger2 = logging.getLogger("module2")
logger2.setLevel(logging.INFO)
logging.basicConfig(
  level=logging.DEBUG,
  format="%(asctime)s - %(name)s - %(levelname)s - %(message)s ",
  datefmt=' %Y-%m-%d  %H:%M:%S'
)
logger1.debug("this is a debug message from logger1")
logger1.error("this is an error message from logger1")
logger2.info("this is an info message from logger2")
```
